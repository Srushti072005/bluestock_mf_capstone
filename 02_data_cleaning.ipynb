```python
import pandas as pd
import numpy as np
import sqlite3
from sqlalchemy import create_engine

print("✅ Libraries imported!")

nav_history = pd.read_csv("data/raw/nav_history.csv")
fund_master = pd.read_csv("data/raw/fund_master.csv")
live_nav = pd.read_csv("data/raw/live_nav.csv")

print("✅ Files loaded!")
print("nav_history shape:", nav_history.shape)
print("fund_master shape:", fund_master.shape)
```

```python
print("=== CLEANING NAV HISTORY ===")

# Parse dates
nav_history['date'] = pd.to_datetime(nav_history['date'])

# Sort by amfi_code + date
nav_history = nav_history.sort_values(['amfi_code', 'date'])

# Remove duplicates
nav_history = nav_history.drop_duplicates()

# Validate NAV > 0
nav_history = nav_history[nav_history['nav'] > 0]

# Forward fill missing NAV
nav_history = nav_history.set_index('date')
nav_history['nav'] = nav_history.groupby('amfi_code')['nav'].ffill()
nav_history = nav_history.reset_index()

# Save cleaned file
nav_history.to_csv("data/processed/nav_history_clean.csv", index=False)

print("✅ nav_history cleaned!")
print("Shape:", nav_history.shape)
print(nav_history.head())
```

```python
# Clean nav_history.csv
print("=== CLEANING NAV HISTORY ===")

# Parse dates
nav_history['date'] = pd.to_datetime(nav_history['date'])

# Sort by scheme_code + date
nav_history = nav_history.sort_values(['scheme_code', 'date'])

# Remove duplicates
nav_history = nav_history.drop_duplicates()

# Validate NAV > 0
nav_history = nav_history[nav_history['nav'] > 0]

# Forward fill missing NAV
nav_history = nav_history.set_index('date')
nav_history['nav'] = nav_history.groupby('scheme_code')['nav'].ffill()
nav_history = nav_history.reset_index()

# Save cleaned file
nav_history.to_csv("data/processed/nav_history_clean.csv", index=False)

print("✅ nav_history cleaned!")
print("Shape:", nav_history.shape)
print(nav_history.head())
```

```python
# Clean nav_history.csv
print("=== CLEANING NAV HISTORY ===")

# Parse dates
nav_history['Date'] = pd.to_datetime(nav_history['Date'])

# Sort by Code + Date
nav_history = nav_history.sort_values(['Code', 'Date'])

# Remove duplicates
nav_history = nav_history.drop_duplicates()

# Validate NAV > 0
nav_history = nav_history[nav_history['NAV'] > 0]

# Forward fill missing NAV
nav_history = nav_history.set_index('Date')
nav_history['NAV'] = nav_history.groupby('Code')['NAV'].ffill()
nav_history = nav_history.reset_index()

# Save cleaned file
import os
os.makedirs("data/processed", exist_ok=True)
nav_history.to_csv("data/processed/nav_history_clean.csv", index=False)

print("✅ nav_history cleaned!")
print("Shape:", nav_history.shape)
print(nav_history.head())
```

```python
# Clean nav_history.csv
print("=== CLEANING NAV HISTORY ===")

# Parse dates with dayfirst=True
nav_history['Date'] = pd.to_datetime(
    nav_history['Date'], 
    dayfirst=True, 
    errors='coerce'
)

# Sort by Code + Date
nav_history = nav_history.sort_values(['Code', 'Date'])

# Remove duplicates
nav_history = nav_history.drop_duplicates()

# Validate NAV > 0
nav_history = nav_history[nav_history['NAV'] > 0]

# Forward fill missing NAV
nav_history = nav_history.set_index('Date')
nav_history['NAV'] = nav_history.groupby('Code')['NAV'].ffill()
nav_history = nav_history.reset_index()

# Save cleaned file
import os
os.makedirs("data/processed", exist_ok=True)
nav_history.to_csv("data/processed/nav_history_clean.csv", index=False)

print("✅ nav_history cleaned!")
print("Shape:", nav_history.shape)
print(nav_history.head())
```

```python
# Clean fund_master.csv
print("=== CLEANING FUND MASTER ===")

# Remove duplicates
fund_master = fund_master.drop_duplicates()

# Remove null values
fund_master = fund_master.dropna()

# Save cleaned file
fund_master.to_csv("data/processed/fund_master_clean.csv", index=False)

print("✅ fund_master cleaned!")
print("Shape:", fund_master.shape)
print(fund_master.head(10))
```

```python
# Clean live_nav.csv
print("=== CLEANING LIVE NAV ===")

# Parse dates
live_nav['Date'] = pd.to_datetime(
    live_nav['Date'], 
    dayfirst=True, 
    errors='coerce'
)

# Remove duplicates
live_nav = live_nav.drop_duplicates()

# Validate NAV > 0
live_nav = live_nav[live_nav['NAV'] > 0]

# Save cleaned file
live_nav.to_csv("data/processed/live_nav_clean.csv", index=False)

print("✅ live_nav cleaned!")
print("Shape:", live_nav.shape)
print(live_nav.head())
```

```python
# Create SQLite Database
print("=== CREATING SQLITE DATABASE ===")

import sqlite3
import os

os.makedirs("data/db", exist_ok=True)

# Create database connection
engine = create_engine('sqlite:///data/db/bluestock_mf.db')

# Save all tables to SQLite
nav_history.to_sql('fact_nav', engine, if_exists='replace', index=False)
fund_master.to_sql('dim_fund', engine, if_exists='replace', index=False)
live_nav.to_sql('fact_live_nav', engine, if_exists='replace', index=False)

print("✅ Database created!")
print("✅ Tables loaded!")

# Verify tables
conn = sqlite3.connect('data/db/bluestock_mf.db')
cursor = conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table'")
tables = cursor.fetchall()
print("\nTables in database:")
for table in tables:
    print(" -", table[0])
conn.close()
```

```python
# Write 10 SQL Queries
print("=== SQL QUERIES ===")

conn = sqlite3.connect('data/db/bluestock_mf.db')

# Query 1 - Top 5 funds by latest NAV
print("\n1. TOP 5 FUNDS BY NAV:")
q1 = pd.read_sql("""
    SELECT Scheme, Code, MAX(NAV) as Max_NAV 
    FROM fact_nav 
    GROUP BY Code 
    ORDER BY Max_NAV DESC 
    LIMIT 5
""", conn)
print(q1)

# Query 2 - Average NAV per scheme
print("\n2. AVERAGE NAV PER SCHEME:")
q2 = pd.read_sql("""
    SELECT Scheme, Code, ROUND(AVG(NAV),2) as Avg_NAV 
    FROM fact_nav 
    GROUP BY Code
""", conn)
print(q2)

# Query 3 - Latest NAV for each scheme
print("\n3. LATEST NAV FOR EACH SCHEME:")
q3 = pd.read_sql("""
    SELECT Scheme, Code, Date, NAV 
    FROM fact_nav 
    WHERE Date = (SELECT MAX(Date) FROM fact_nav)
""", conn)
print(q3)

# Query 4 - Total records per scheme
print("\n4. TOTAL RECORDS PER SCHEME:")
q4 = pd.read_sql("""
    SELECT Scheme, Code, COUNT(*) as Total_Records 
    FROM fact_nav 
    GROUP BY Code
""", conn)
print(q4)

# Query 5 - All fund names
print("\n5. ALL FUND NAMES:")
q5 = pd.read_sql("""
    SELECT * FROM dim_fund
""", conn)
print(q5)

conn.close()
print("\n✅ SQL Queries Complete!")
```

```python
# 5 More SQL Queries
conn = sqlite3.connect('data/db/bluestock_mf.db')

# Query 6 - Minimum NAV per scheme
print("\n6. MINIMUM NAV PER SCHEME:")
q6 = pd.read_sql("""
    SELECT Scheme, Code, ROUND(MIN(NAV),2) as Min_NAV 
    FROM fact_nav 
    GROUP BY Code
""", conn)
print(q6)

# Query 7 - Maximum NAV per scheme
print("\n7. MAXIMUM NAV PER SCHEME:")
q7 = pd.read_sql("""
    SELECT Scheme, Code, ROUND(MAX(NAV),2) as Max_NAV 
    FROM fact_nav 
    GROUP BY Code
""", conn)
print(q7)

# Query 8 - NAV trend last 7 days
print("\n8. NAV LAST 7 DAYS:")
q8 = pd.read_sql("""
    SELECT Scheme, Code, Date, NAV 
    FROM fact_nav 
    ORDER BY Date DESC 
    LIMIT 7
""", conn)
print(q8)

# Query 9 - Compare live vs history NAV
print("\n9. LIVE NAV COMPARISON:")
q9 = pd.read_sql("""
    SELECT Scheme, Code, NAV 
    FROM fact_live_nav
""", conn)
print(q9)

# Query 10 - Fund master with NAV count
print("\n10. FUND MASTER WITH NAV COUNT:")
q10 = pd.read_sql("""
    SELECT d.schemeName, COUNT(f.NAV) as NAV_Count
    FROM dim_fund d
    LEFT JOIN fact_nav f ON d.schemeCode = f.Code
    GROUP BY d.schemeName
""", conn)
print(q10)

conn.close()
print("\n✅ All 10 SQL Queries Complete!")
```
